```python
import math
from typing import Sequence, Tuple


def mean(values: Sequence[float]) -> float:
    """Выборочное среднее арифметическое: sum(x_i) / n."""
    return sum(values) / len(values)


def covariance(xs: Sequence[float], ys: Sequence[float]) -> float:
    """
    Выборочная ковариация с делением на n:
        cov = (1/n) * Σ (x_i - x̄)(y_i - ȳ)
    Эквивалентная формула: cov = (1/n) * Σ x_i y_i - x̄ * ȳ
    """
    n = len(xs)
    mx, my = mean(xs), mean(ys)
    return sum((x - mx) * (y - my) for x, y in zip(xs, ys)) / n


def pearson(xs: Sequence[float], ys: Sequence[float]) -> float:
    """
    Коэффициент линейной корреляции Пирсона:
        r = cov(x, y) / (σ_x * σ_y),  где σ — среднеквадратичные отклонения
        σ_x = sqrt((1/n) * Σ (x_i - x̄)^2)
    Если хотя бы одна из дисперсий равна нулю, корреляция не определена —
    возвращаем nan.
    """
    n = len(xs)
    mx, my = mean(xs), mean(ys)
    cov = sum((x - mx) * (y - my) for x, y in zip(xs, ys)) / n
    var_x = sum((x - mx) ** 2 for x in xs) / n
    var_y = sum((y - my) ** 2 for y in ys) / n
    if var_x == 0 or var_y == 0:
        return float("nan")
    return cov / math.sqrt(var_x * var_y)


def correlation(xs: Sequence[float], ys: Sequence[float]) -> Tuple[float, float]:
    """
    Возвращает кортеж (cov, corr):
      cov  — выборочная ковариация (с делением на n);
      corr — коэффициент линейной корреляции Пирсона.
    Проверяет, что длины списков совпадают и n >= 2.
    """
    if len(xs) != len(ys):
        raise ValueError("Списки xs и ys должны иметь одинаковую длину.")
    if len(xs) < 2:
        raise ValueError("Длина списков должна быть n >= 2.")
    return covariance(xs, ys), pearson(xs, ys)
```

Демонстрация


```python
if __name__ == "__main__":
    # Пример из условия
    xs1 = [1, 2, 3, 4]
    ys1 = [2, 4, 6, 8]
    cov1, corr1 = correlation(xs1, ys1)
    print(f"correlation({xs1}, {ys1}) = ({cov1}, {corr1})")
    assert math.isclose(cov1, 2.5) and math.isclose(corr1, 1.0)

    # Пример с отрицательной корреляцией
    xs2 = [1, 2, 3, 4, 5]
    ys2 = [10, 8, 6, 4, 2]
    cov2, corr2 = correlation(xs2, ys2)
    print(f"correlation({xs2}, {ys2}) = ({cov2}, {corr2})")

    # Пример с некоррелированными данными
    xs3 = [-2, -1, 0, 1, 2]
    ys3 = [4, 1, 0, 1, 4]  # квадратичная зависимость -> r ≈ 0
    cov3, corr3 = correlation(xs3, ys3)
    print(f"correlation({xs3}, {ys3}) = ({cov3}, {corr3})")
```
